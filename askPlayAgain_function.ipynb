````python
def ask_play_again():
    """
    Questa funzione, alla fine di una partita, chiede all'utente se vuole giocare di nuovo.

    ---------
    Parametri: nessuno.

    ---------
    Logica di funzionamento:
    1. Chiede all'utente se vuole fare un'altra partita tramite input().
    2. Pulizia dell'input: rimuove spazi accidentali e converte in minuscolo (.strip().lower()).
    3. Se la risposta è affermativa ("s", "si", "sì") restituisce True, se è negativa ("n", "no") restituisce False.
    4. Se la risposta non è valida stampa un messaggio di errore e ripete la domanda (ciclo while).

    ---------
    Restituisce: True se l'utente vuole giocare un'altra partita, False se vuole uscire dal gioco.

    ---------
    Deve essere richiamata alla fine di ogni partita, cioè quando flag diventa 1 (vittoria) o 2 (sconfitta).
    Se restituisce True il programma principale ricomincia dalla scelta della difficoltà.
    """
    while True:  # la domanda viene ripetuta finché l'utente non dà una risposta valida
        answer = input("Vuoi giocare di nuovo? (s/n): ").strip().lower()

        if answer in ["s", "si", "sì"]:
            return True
        elif answer in ["n", "no"]:
            return False
        else:
            print("Input non valido: rispondi s (sì) oppure n (no).")
````

#### TEST

````python
# Supponiamo che la partita sia finita e che l'utente voglia rigiocare: eseguire la cella e inserire "Sì" quando richiesto.

play = ask_play_again()
print(f"Vuole rigiocare: {play}\n")

# Ci si aspetta: Vuole rigiocare: True ("Sì" diventa "sì" grazie a .lower())
````

````python
# Supponiamo che la partita sia finita e che l'utente non voglia rigiocare: eseguire la cella e inserire "n" quando richiesto.

play = ask_play_again()
print(f"Vuole rigiocare: {play}\n")

# Ci si aspetta: Vuole rigiocare: False
````

````python
# Supponiamo che l'utente inserisca una risposta non valida: eseguire la cella e inserire prima "boh", poi una riga vuota (solo Invio) e infine "no".

play = ask_play_again()
print(f"Vuole rigiocare: {play}\n")

# Ci si aspetta un messaggio di errore per "boh" e per la riga vuota, con la domanda ripetuta ogni volta,
# e infine: Vuole rigiocare: False
````
