```python
def ask_to_play():
    """Se l'utente vuole cominciare a giocare: ``True`` per sì, ``False`` per no.

    La domanda si ripete finché la risposta non è una di quelle riconosciute.
    Una risposta sbagliata non costa una vita: la partita non è ancora
    cominciata, e non ci sono vite da togliere.

    **Risposte riconosciute.** ``s``, ``si`` e ``sì`` per il sì, ``n`` e ``no``
    per il no. Maiuscole e spazi ai lati non contano: ``" Sì "`` vale ``"sì"``.

    **Chi la usa.** Il codice principale, una volta sola, all'avvio. Con
    ``True`` si passa alla scelta della difficoltà (``choose_difficulty``); con
    ``False`` il programma termina senza giocare.
    """
    while True:
        risposta = input("Vuoi giocare all'impiccato? (s/n): ").strip().lower()

        if risposta in ["s", "si", "sì"]:
            return True
        if risposta in ["n", "no"]:
            return False

        # Qui si arriva solo con una risposta non riconosciuta. Il ciclo
        # ripete la domanda: l'unica uscita è uno dei due return qui sopra.
        print("Input non valido: rispondi s (sì) oppure n (no).")
```

## Verifiche
Ogni cella chiede una risposta da tastiera: va inserita quella indicata sopra la cella.

### Un sì restituisce True
**Obiettivo**: Verificare che la risposta affermativa più corta sia riconosciuta.

**Da inserire**: `s`

**Atteso**: `ask_to_play() -> True`

```python
gioca = ask_to_play()
print("ask_to_play() ->", gioca)
```

### Un no in maiuscolo vale come in minuscolo
**Obiettivo**: Verificare che `.lower()` renda indifferente il modo in cui l'utente scrive la risposta.

**Da inserire**: `NO`

**Atteso**: `ask_to_play() -> False`

```python
gioca = ask_to_play()
print("ask_to_play() ->", gioca)
```

### Una risposta non riconosciuta fa ripetere la domanda
**Obiettivo**: Verificare che una risposta sbagliata, anche un numero, non chiuda la funzione e non la faccia andare in errore, ma riproponga la domanda finché non arriva una risposta valida.

**Da inserire**: `forse`, poi `5`, poi `si`

**Atteso**: il messaggio di errore due volte, poi `ask_to_play() -> True`

```python
gioca = ask_to_play()
print("ask_to_play() ->", gioca)
```
