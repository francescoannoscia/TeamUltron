```python
def ask_to_play():
    """
    Questa funzione chiede all'utente se vuole iniziare a giocare all'impiccato.

    ---------
    Parametri: nessuno.

    ---------
    Logica di funzionamento:
    1. Chiede all'utente se vuole giocare tramite input().
    2. Pulizia dell'input: rimuove spazi accidentali e converte in minuscolo (.strip().lower()),
       così "S", " si " e "Sì" vengono accettati allo stesso modo.
    3. Se la risposta è affermativa ("s", "si", "sì") restituisce True, se è negativa ("n", "no") restituisce False.
    4. Se la risposta non è valida stampa un messaggio di errore e ripete la domanda (ciclo while).
       Non viene tolta nessuna vita, perché la partita non è ancora iniziata.

    ---------
    Restituisce: True se l'utente vuole giocare, False altrimenti.

    ---------
    Deve essere richiamata una sola volta, all'avvio del programma, prima di scegliere la difficoltà.
    """
    while True:  # la domanda viene ripetuta finché l'utente non dà una risposta valida
        answer = input("Vuoi giocare all'impiccato? (s/n): ").strip().lower()

        if answer in ["s", "si", "sì"]:
            return True  # il return interrompe sia il ciclo sia la funzione
        elif answer in ["n", "no"]:
            return False
        else:
            print("Input non valido: rispondi s (sì) oppure n (no).")
```

#### TEST

```python
# Supponiamo che l'utente voglia giocare: eseguire la cella e inserire "s" quando richiesto.

play = ask_to_play()
print(f"Vuole giocare: {play}\n")

# Ci si aspetta: Vuole giocare: True
```

```python
# Supponiamo che l'utente non voglia giocare: eseguire la cella e inserire "NO" (in maiuscolo) quando richiesto.

play = ask_to_play()
print(f"Vuole giocare: {play}\n")

# Ci si aspetta: Vuole giocare: False (il maiuscolo viene convertito da .lower())
```

```python
# Supponiamo che l'utente inserisca una risposta non valida: eseguire la cella e inserire prima "forse", poi "5" e infine "si".

play = ask_to_play()
print(f"Vuole giocare: {play}\n")

# Ci si aspetta un messaggio di errore per "forse" e per "5", con la domanda ripetuta ogni volta,
# e infine: Vuole giocare: True
```
