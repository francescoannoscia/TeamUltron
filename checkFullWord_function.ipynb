```python
# Funzione presente nel file "handleError_function"

def handle_error(lives, flag):
    print("Errore!")
    lives -= 1
    
    if lives <= 0:
        flag = 2  # 2 indica la sconfitta
        
    return lives, flag
```

```python
def check_full_word(user_input, secret_word, lives, flag):
    """
    Questa funzione gestisce il tentativo dell'utente di indovinare l'intera parola.

    ---------
    Parametri: user_input (la parola tentata), secret_word (la parola da indovinare), 
    lives (i tentativi rimasti) e flag (lo stato della partita).
    
    ---------
    Restituisce: le 2 variabili modificate (lives e flag) e lo stato del sinfolo turno (che indica se l'utente ha indovinato o meno la parola)
    nel punto del codice in cui la funzione è stata richiamata.

    ---------
    Deve essere richiamata solo quando l'utente inserisce una stringa lunga 2 o più caratteri 
    (ovvero quando la condizione lunghezza dell'input == 1 è falsa).
    """
    word_found = False
    if user_input == secret_word:
        word_found = True  #variabile per tracciare solo l'esito del turno
        print("Hai indovinato la parola!")
        flag = 1  # 1 indica la vittoria
        return lives, flag, word_found
    else:
        print("Parola errata!")
        lives, flag = handle_error(lives, flag)
        return lives, flag, word_found
```

#### TEST

```python
# Supponiamo che l'utente abbia indovinato la parola segreta

lives = 5
flag = 0
lives, flag, word_found = check_full_word("chitarra", "chitarra", lives, flag)
print(f"Vite rimaste: {lives}, Stato Flag: {flag}, Stato turno: {word_found}\n")

# Ci si aspetta un messaggio di successo, che le vite siano rimaste intatte e che lo stato della partita sia vittoria (flag = 1)
```

```python
# Supponiamo che l'utente non abbia indovinato la parola segreta

lives = 5
flag = 0
lives, flag, word_found = check_full_word("batteria", "chitarra", lives, flag)
print(f"Vite rimaste: {lives}, Stato Flag: {flag}, Stato turno: {word_found}\n")

# Ci si aspetta un messaggio di errore, che venga sottratta una vita e che lo stato della partita sia in corso (flag = 0)
```

```python
# Supponiamo che l'utente non abbia indovinato la parola segreta e che gli sia rimasta una sola vita

lives = 1
flag = 0
lives, flag, word_found = check_full_word("pianoforte", "chitarra", lives, flag)
print(f"Vite rimaste: {lives}, Stato Flag: {flag}, Stato turno: {word_found}\n")

# Ci si aspetta un messaggio di errore, che venga sottratta l'ultima vita e che lo stato della partita sia sconfitta (flag = 2)
```
