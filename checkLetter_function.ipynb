```python
def check_letter(l):

    """
    Questa funzione prende in input una lettera e verifica se è una consonante o una vocale: se consonante, la restituisce tale; se vocale,
    restituisce una lista che contiene sé stessa e le possibili forme accentate presenti nelle parole italiane.
    ---------
    Parameters:
    l, str
        Lettera
    ---------
    Returns:
    l, str/list
        Lettera (str) se consonante oppure lista di alcune forme accentate della lettera (list) se vocale
    ---------
    Richiede verifica di input ('l' deve essere un carattere alfabetico).
    """

    if l in "aeiou":
        if l == "a":
            l = ["a","à"]
        elif l == "e":
            l = ["e","è","é"]
        elif l == "i":
            l == ["i","ì"]
        elif l == "o":
            l = ["o","ò"]
        elif l == "u":
            l = ["u","ù"]

    return l
```

```python
check_letter("t")
```

```python
check_letter("e")
```
