```python
# Funzione presente nel file "checkLetter_function"

def check_letter(l):

    if l in "aeiou":
        if l == "a":
            l = ["a","à"]
        elif l == "e":
            l = ["e","è","é"]
        elif l == "i":
            l = ["i","ì"]
        elif l == "o":
            l = ["o","ò"]
        elif l == "u":
            l = ["u","ù"]
    else: l = [l]

    return l
```

```python
# Funzione presente nel file "handleError_function"

def handle_error(lives, flag):
    print("Errore!")
    lives -= 1
    
    if lives <= 0:
        flag = 2  # 2 indica la sconfitta
        
    return lives, flag
```

```python
def check_single_letter(user_input, secret_word, lives, flag):
    """
    Questa funzione gestisce il tentativo dell'utente di indovinare una singola lettera, verificando 
    sia la lettera base che le sue eventuali varianti accentate.

    ---------
    Parametri:
    - user_input: la lettera singola inserita dall'utente.
    - secret_word: la parola segreta da indovinare.
    - lives: il numero di vite/tentativi rimasti.
    - flag: lo stato attuale della partita (0 = in corso, 1 = vittoria, 2 = sconfitta).

    ---------
    Logica di funzionamento:
    1. Acquisizione varianti: Richiama la funzione check_letter per acquisire le varianti (sotto forma di lista) della lettera 
    inserita dall'utente (es. "b" -> ["b"] e "a" -> ["a", "à"]).
    2. Ricerca nella parola: Inizializza un interruttore 'letter_found' a False. Un ciclo 'for' passa in rassegna le varianti. 
    Se una variante è presente nella parola segreta, 'letter_found' diventa True e un 'break' interrompe immediatamente il ciclo.
    3. Gestione dell'errore: Utilizza un costrutto 'for...else'. Se il ciclo termina senza che nessuna variante sia stata trovata 
    (letter_found rimasto false), l'else richiama handle_error per sottrarre una vita. L'allineamento al 'for' garantisce che la 
    vita venga sottratta solo dopo aver verificato tutte le varianti.

    ---------
    Restituisce:
    I valori aggiornati di 'lives' e 'flag', la lista 'letter_variants' e l'interruttore 'letter_found'. Questi ultimi due dati 
    verranno passati al ciclo principale per sostituire graficamente i trattini corretti a schermo.
    """
        
    # Richiamiamo la funzione check_letter per ottenere le varianti (es: se 'e' -> ['e', 'è', 'é']) (restituisce una lista)
    letter_variants = check_letter(user_input)
        
    # Verifichiamo se almeno una delle varianti (es: 'a' o 'à') è presente nella parola segreta
    letter_found = False
    for let in letter_variants:
        if let in secret_word:
            letter_found = True
            break # Basta trovarne una per considerare il tentativo corretto
            
    else:
        lives, flag = handle_error(lives, flag) # Se letter_found è rimasto false, la lettera non è presente nella parola segreta e 
                                                # richiamiamo la funzione per la gestione delle vite
        
    # Restituiamo anche 'letter_variants' (es: ['a', 'à']) così il ciclo principale 
    # sa quali esatte lettere andare a svelare nella stringa nascosta.
    return lives, flag, letter_variants, letter_found
```
