```python
def choose_difficulty():
    """La difficoltà scelta dall'utente, nella forma che ``random_word`` si
    aspetta: ``"facile"`` o ``"difficile"``.

    L'utente risponde con un numero, ``1`` o ``2``; la funzione restituisce la
    parola. Così il resto del codice non deve sapere quale numero corrisponde a
    quale livello: lo sa solo questa funzione.

    **Perché il confronto è fra stringhe.** ``input`` restituisce sempre una
    stringa. Convertirla con ``int`` fermerebbe il programma con un
    ``ValueError`` alla prima lettera digitata; confrontarla con ``"1"`` e
    ``"2"`` tratta una lettera come qualunque altra risposta sbagliata.

    **Chi la usa.** Il codice principale, all'inizio di ogni partita e prima
    di ``random_word``: ``secret_word = random_word(choose_difficulty())``.
    """
    while True:
        # Solo strip, senza lower: le risposte valide sono cifre, e una cifra
        # non ha maiuscole.
        scelta = input("Scegli la difficoltà (1 = facile, 2 = difficile): ").strip()

        if scelta == "1":
            return "facile"
        if scelta == "2":
            return "difficile"

        # Una risposta diversa da 1 e 2 non costa una vita: la partita non è
        # ancora cominciata. Si ripete la domanda.
        print("Input non valido: inserisci 1 oppure 2.")
```

## Verifiche
Ogni cella chiede una risposta da tastiera: va inserita quella indicata sopra la cella.

### 1 corrisponde alla modalità facile
**Obiettivo**: Verificare che la funzione traduca il numero nella parola che `random_word` riconosce.

**Da inserire**: `1`

**Atteso**: `choose_difficulty() -> facile`

```python
livello = choose_difficulty()
print("choose_difficulty() ->", livello)
```

### Gli spazi intorno al numero non contano
**Obiettivo**: Verificare che uno spazio digitato per sbaglio prima o dopo il numero venga tolto da `.strip()`.

**Da inserire**: `␣2␣` (il 2 con uno spazio prima e uno dopo)

**Atteso**: `choose_difficulty() -> difficile`

```python
livello = choose_difficulty()
print("choose_difficulty() ->", livello)
```

### Un numero fuori scelta o una parola fanno ripetere la domanda
**Obiettivo**: Verificare che né un numero diverso da 1 e 2 né il nome del livello scritto per esteso siano accettati, e che il programma non vada in errore.

**Da inserire**: `3`, poi `facile`, poi `1`

**Atteso**: il messaggio di errore due volte, poi `choose_difficulty() -> facile`

```python
livello = choose_difficulty()
print("choose_difficulty() ->", livello)
```
