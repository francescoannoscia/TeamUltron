```python
def choose_difficulty():
    """
    Questa funzione chiede all'utente di scegliere il livello di difficoltà del gioco.

    ---------
    Parametri: nessuno.

    ---------
    Logica di funzionamento:
    1. Chiede all'utente di inserire 1 (facile) oppure 2 (difficile) tramite input().
    2. Pulizia dell'input: rimuove eventuali spazi accidentali (.strip()).
    3. L'input viene confrontato con le stringhe "1" e "2" e non convertito in numero con int():
       in questo modo, se l'utente scrive una lettera, il programma non va in errore.
    4. Se l'input non è valido stampa un messaggio di errore e ripete la domanda (ciclo while).
       Non viene tolta nessuna vita, perché la partita non è ancora iniziata.

    ---------
    Restituisce: la stringa "facile" oppure "difficile", cioè il formato richiesto dalla funzione random_word.

    ---------
    Deve essere richiamata dopo che l'utente ha deciso di giocare e prima di generare la parola segreta:
    difficulty = choose_difficulty()
    secret_word = random_word(difficulty)
    """
    while True:  # la domanda viene ripetuta finché l'utente non inserisce 1 o 2
        choice = input("Scegli la difficoltà (1 = facile, 2 = difficile): ").strip()

        if choice == "1":
            return "facile"
        elif choice == "2":
            return "difficile"
        else:
            print("Input non valido: inserisci 1 oppure 2.")
```

#### TEST

```python
# Supponiamo che l'utente scelga la modalità facile: eseguire la cella e inserire "1" quando richiesto.

difficulty = choose_difficulty()
print(f"Difficoltà scelta: {difficulty}\n")

# Ci si aspetta: Difficoltà scelta: facile
```

```python
# Supponiamo che l'utente scelga la modalità difficile: eseguire la cella e inserire " 2 " (con degli spazi) quando richiesto.

difficulty = choose_difficulty()
print(f"Difficoltà scelta: {difficulty}\n")

# Ci si aspetta: Difficoltà scelta: difficile (gli spazi vengono rimossi da .strip())
```

```python
# Supponiamo che l'utente inserisca valori non validi: eseguire la cella e inserire prima "3", poi "facile" e infine "1".

difficulty = choose_difficulty()
print(f"Difficoltà scelta: {difficulty}\n")

# Ci si aspetta un messaggio di errore per "3" e per "facile", con la domanda ripetuta ogni volta,
# e infine: Difficoltà scelta: facile
```
