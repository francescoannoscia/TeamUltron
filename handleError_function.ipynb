```python
def handle_error(lives, flag):
    """
    Questa funzione gestisce le penalità (le vite) in caso di errore di input. Se la condizione lives <= 0 è vera il giocatore ha esaurito i tentativi,
    il flag viene impostato a 2 cambiando lo stato del gioco a sconfitta.

    ---------
    Parametri: la variabile lives (i tentativi rimasti) e la variabile flag (per tracciare lo stato della partita: in corso, vinta o persa).
    
    ---------
    Restituisce: le 2 variabili modificate (vite e flag) nel punto del codice in cui la funzione è stata richiamata.

    ---------
   Deve essere richiamata (lives, flag = gestisci_errore(lives, flag)) ogni volta che l'utente commette uno sbaglio che gli costa una vita. 
    Ovvero:
    - Input non valido: quando l'utente inserisce numeri, spazi o caratteri speciali.
    - Tentativo di parola intera fallito: quando la parola inserita non è uguale alla parola segreta.
    - Lettera singola errata: quando la lettera (e le sue varianti accentate) non è presente 
      nella parola segreta.
    """
    print("Errore!")
    lives -= 1
    
    if lives <= 0:
        flag = 2  # 2 indica la sconfitta
        
    return lives, flag
```

#### TEST

```python
# Simuliamo di avere 5 vite e flag a 0 (partita in corso)

lives = 5
flag = 0
lives, flag = handle_error(lives, flag)
print(f"Vite rimaste: {lives}, Stato Flag: {flag}")

# Poichè è stato commesso un errore, ci si aspetta che venga sottratta una vita e, poichè rimangono ancora 4 vite, che la partita sia ancora in corso.
```
